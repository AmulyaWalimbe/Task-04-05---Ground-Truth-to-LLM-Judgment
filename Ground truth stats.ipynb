```python
import csv

def load_csv(path):
    with open(path, newline="", encoding="utf-8") as f:
        return list(csv.DictReader(f))


def main():
    games = load_csv("D:/opt research work/task 04 & 05/game_log_2025.csv")
    players = load_csv("D:/opt research work/task 04 & 05/player_season_stats_2025.csv")

    print("=" * 60)
    print("GROUND TRUTH: 2025 Syracuse Women's Lacrosse Season")
    print("=" * 60)

    #Basic season facts
    total_games = len(games)
    wins = [g for g in games if g["result"] == "W"]
    losses = [g for g in games if g["result"] == "L"]
    print(f"\nGames played: {total_games}")
    print(f"Record: {len(wins)}-{len(losses)}")

    #Top scorer
    top_scorer = max(players, key=lambda p: int(p["goals"]))
    print(f"\nMost goals: {top_scorer['player']} ({top_scorer['goals']} goals)")

    top_points = max(players, key=lambda p: int(p["points"]))
    print(f"Most points: {top_points['player']} ({top_points['points']} points)")

    top_assists = max(players, key=lambda p: int(p["assists"]))
    print(f"Most assists: {top_assists['player']} ({top_assists['assists']} assists)")

    #Average margin of victory in wins
    margins = [int(g["syracuse_score"]) - int(g["opponent_score"]) for g in wins]
    avg_margin = sum(margins) / len(margins)
    print(f"\nAverage margin of victory in wins: {avg_margin:.2f}")
    print(f"  (margins: {margins})")

    #Average margin of defeat in losses
    loss_margins = [int(g["opponent_score"]) - int(g["syracuse_score"]) for g in losses]
    avg_loss_margin = sum(loss_margins) / len(loss_margins)
    print(f"Average margin of defeat in losses: {avg_loss_margin:.2f}")

    #Highest combined score game
    combined = [
        (int(g["syracuse_score"]) + int(g["opponent_score"]), g)
        for g in games
    ]
    highest = max(combined, key=lambda x: x[0])
    print(
        f"\nHighest combined score: {highest[0]} "
        f"(SU {highest[1]['syracuse_score']} - "
        f"{highest[1]['opponent_score']} {highest[1]['opponent']}, "
        f"{highest[1]['date']})"
    )

    #Lowest combined score game
    lowest = min(combined, key=lambda x: x[0])
    print(
        f"Lowest combined score: {lowest[0]} "
        f"(SU {lowest[1]['syracuse_score']} - "
        f"{lowest[1]['opponent_score']} {lowest[1]['opponent']}, "
        f"{lowest[1]['date']})"
    )

    #Biggest win and biggest loss
    biggest_win = max(wins, key=lambda g: int(g["syracuse_score"]) - int(g["opponent_score"]))
    print(
        f"\nBiggest win: {biggest_win['syracuse_score']}-{biggest_win['opponent_score']} "
        f"vs {biggest_win['opponent']} ({biggest_win['date']})"
    )
    biggest_loss = max(losses, key=lambda g: int(g["opponent_score"]) - int(g["syracuse_score"]))
    print(
        f"Biggest loss: {biggest_loss['syracuse_score']}-{biggest_loss['opponent_score']} "
        f"vs {biggest_loss['opponent']} ({biggest_loss['date']})"
    )

    #Season totals
    total_goals = sum(int(g["syracuse_score"]) for g in games)
    total_opp_goals = sum(int(g["opponent_score"]) for g in games)
    print(f"\nTotal goals scored (season): {total_goals}")
    print(f"Total goals allowed (season): {total_opp_goals}")

    player_goal_sum = sum(int(p["goals"]) for p in players)
    print(f"Sum of individual player goals: {player_goal_sum}")
    if player_goal_sum != total_goals:
        print(
            f"  NOTE: this is {total_goals - player_goal_sum} less than the team "
            f"total ({total_goals}) reported on the official stats page -- a "
            f"data quality quirk present in the source itself, not an error in "
            f"this script."
        )

    #Phase B: a defined "game changer" metric
    #Definition: goals + assists + (2 x game-winning goals).
    #The x2 weight on game-winning goals reflects that a GWG has a
    #direct, provable impact on a win, not just raw scoring volume.
    print("\n--- Phase B metric: GAME CHANGER SCORE ---")
    print("Definition: goals + assists + (2 x game_winning_goals)")
    scored = []
    for p in players:
        score = int(p["goals"]) + int(p["assists"]) + 2 * int(p["game_winning_goals"])
        scored.append((score, p["player"]))
    scored.sort(reverse=True)
    for score, name in scored[:5]:
        print(f"  {name:25s} {score}")


if __name__ == "__main__":
    main()
```
